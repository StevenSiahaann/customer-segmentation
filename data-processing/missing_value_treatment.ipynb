```python
import pandas as pd
import numpy as np
```

```python
data_pendapatan=pd.read_csv('../data/all_data_pendapatan_mei_november_2023.csv').drop(columns=['Unnamed: 0'])
```

```python
data_pendapatan
```

# Missing Value

```python
missing_data = data_pendapatan.isnull().sum()
missing_percentage = (missing_data[missing_data > 0] / data_pendapatan.shape[0]) * 100
missing_percentage.sort_values(ascending=True, inplace=True)
```

```python
missing_percentage
```

```python
data_pendapatan_col_cleared=data_pendapatan.drop(columns=['LEVEL_0_EN','LEVEL_1_EN','AP','NAMA_AKUN_EN','LEVEL_2_EN','LEVEL_3_EN'])
data_pendapatan_col_cleared
```

```python
# nama kapal, gt, loa, tanggal pembatalan have a many missing value, so drop the columns
added_columns_to_drop = ['Nama Kapal','GT','LOA','Tanggal Pembatalan']
data_pendapatan_col_cleared=data_pendapatan_col_cleared.drop(columns=added_columns_to_drop)
data_pendapatan_col_cleared
```
