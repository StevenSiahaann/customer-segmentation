# Convert Columns to Relevance Type

```python
data_piutang['Customer'] = data_piutang['Customer'].astype('int64')
data_piutang['Company Code'] = data_piutang['Company Code'].astype('int64')
data_piutang['Profit Center'] = data_piutang['Profit Center'].astype('int64')
data_piutang['Document Number'] = data_piutang['Document Number'].astype('int64')

data_piutang['Fiscal Year'] = data_piutang['Fiscal Year'].astype(int)
data_piutang['Fiscal Year'] = pd.to_datetime(data_piutang['Fiscal Year'], format='%Y')

data_piutang['CoA Code'] = data_piutang['CoA Code'].astype('int64')

```
