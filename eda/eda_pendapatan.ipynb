## Integer Columns ('Pendapatan in USD', 'Pendapatan in IDR','Kontribusi in USD', 'Kontribusi in IDR')

### Insight
1. There are 21698 negative data in 'Pendapatan in IDR' (potentially incorrect data)
2. All values for 'Kontribusi in USD' and 'Kontribusi in IDR' are zero
3. There are 1410771 of 'Pendapatan in USD' have a zero values

```python
columns_to_check = ['Pendapatan in USD', 'Pendapatan in IDR', 'Kontribusi in USD', 'Kontribusi in IDR']

for col in columns_to_check:
    min_val = data_pendapatan[col].min()
    max_val = data_pendapatan[col].max()
    mean_val = data_pendapatan[col].mean()
    null_count = data_pendapatan[col].isnull().sum()    
    negative_count = (data_pendapatan[col] < 0).sum()
    zero_count = (data_pendapatan[col] == 0).sum()

    print(f"Insight untuk kolom '{col}':")
    print(f"  Range Nilai: {min_val} hingga {max_val}")
    print(f"  Rata-rata Nilai: {mean_val}")
    print(f"  Jumlah Nilai Null: {null_count}")
    print(f"  Jumlah Data Negatif: {negative_count}")
    print(f"  Jumlah Data Zero (0): {zero_count}")
```
