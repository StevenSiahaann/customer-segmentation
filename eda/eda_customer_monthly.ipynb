```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
data_customer=pd.read_excel('../data/pendapatan_preprocess_v1-monthly-customer.xlsx').drop(columns=['Unnamed: 0'])
```

# Monthly General Trend 

```python
data_customer['Tanggal Transaksi'] = pd.to_datetime(data_customer['Tanggal Transaksi'], format='%m-%Y')
monthly_freq = data_customer.groupby(
    [pd.Grouper(key='Tanggal Transaksi', freq='M')]
).agg({
    'jumlah_transaksi_pembelian': 'sum',
    'jumlah_transaksi_retur': 'sum'
}).reset_index()
monthly_freq_melted = monthly_freq.melt(id_vars=['Tanggal Transaksi'], 
                                        value_vars=['jumlah_transaksi_pembelian', 'jumlah_transaksi_retur'],
                                        var_name='Tipe Transaksi', 
                                        value_name='Jumlah Transaksi')

plt.figure(figsize=(10, 6))
ax = sns.barplot(x='Tanggal Transaksi', y='Jumlah Transaksi', hue='Tipe Transaksi', 
                 data=monthly_freq_melted, palette='viridis')

for p in ax.patches:
    ax.annotate(f'{p.get_height():,.0f}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                fontsize=10, color='black', 
                xytext=(0, 5), textcoords='offset points')

plt.title('Jumlah Transaksi per Bulan', fontsize=16)
plt.xlabel('Bulan')
plt.ylabel('Jumlah Transaksi')
plt.xticks(rotation=45)
plt.legend(title='Tipe Transaksi')
plt.tight_layout()
plt.show()
```

# Monthly Trend Frequency Per Customer

## Top 10 Customer Based on Pembelian and Retur Each Month

```python
monthly_customer_freq = data_customer.groupby(
    [pd.Grouper(key='Tanggal Transaksi', freq='M'), 'Cust No.','cust_name']
).agg({
    'jumlah_transaksi_pembelian': 'sum',
    'jumlah_transaksi_retur': 'sum'
}).reset_index()
unique_months = monthly_customer_freq['Tanggal Transaksi'].unique()
```

```python
unique_months = monthly_customer_freq['Tanggal Transaksi'].unique()
n_months = len(unique_months)

n_cols = 2
n_rows = int(np.ceil(n_months / n_cols))

plt.style.use('default')
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 7*n_rows))
axes = axes.flatten()

for idx, month in enumerate(sorted(unique_months)):
    month_data = monthly_customer_freq[monthly_customer_freq['Tanggal Transaksi'] == month]
    
    month_data['total_transaksi'] = month_data['jumlah_transaksi_pembelian'] + month_data['jumlah_transaksi_retur']
    top_10 = month_data.nlargest(10, 'total_transaksi')

    ax = axes[idx]
    width = 0.35

    bars1 = ax.barh(np.arange(len(top_10)), 
                    top_10['jumlah_transaksi_pembelian'], 
                    width, 
                    label='Pembelian',
                    color='forestgreen')
    
    bars2 = ax.barh(np.arange(len(top_10)) + width, 
                    top_10['jumlah_transaksi_retur'], 
                    width, 
                    label='Retur',
                    color='crimson')

    ax.set_title(f'Top 10 Customer - {month.strftime("%B %Y")}')
    ax.set_xlabel('Jumlah Transaksi')
    ax.set_yticks(np.arange(len(top_10)) + width/2)
    ax.set_yticklabels(top_10['cust_name'])
    ax.legend()

    def add_labels(bars):
        for bar in bars:
            width = bar.get_width()
            ax.text(width, 
                   bar.get_y() + bar.get_height()/2, 
                   f'{int(width)}', 
                   ha='left', 
                   va='center',
                   fontsize=8)

    add_labels(bars1)
    add_labels(bars2)

if n_months < len(axes):
    for idx in range(n_months, len(axes)):
        fig.delaxes(axes[idx])

plt.tight_layout()
plt.show()
```

# Monthly Nominal Trend

```python
import numpy as np
import matplotlib.pyplot as plt

unique_months = monthly_customer_freq['Tanggal Transaksi'].unique()
n_months = len(unique_months)

n_cols = 2
n_rows = int(np.ceil(n_months / n_cols))

plt.style.use('default')
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 7 * n_rows))
axes = axes.flatten()

for idx, month in enumerate(sorted(unique_months)):
    month_data = monthly_customer_freq[monthly_customer_freq['Tanggal Transaksi'] == month]
    
    top_pembelian = month_data.sort_values('jumlah_transaksi_pembelian', ascending=False).head(1)

    top_retur = month_data.sort_values('jumlah_transaksi_retur', ascending=False).head(1)

    ax = axes[idx]
    width = 0.35

    bars1 = ax.barh([0], 
                    top_pembelian['jumlah_transaksi_pembelian'], 
                    width, 
                    label=f'Pembelian Terbesar: {top_pembelian["cust_name"].values[0]}',
                    color='forestgreen')
    
    bars2 = ax.barh([1], 
                    top_retur['jumlah_transaksi_retur'], 
                    width, 
                    label=f'Retur Terbesar: {top_retur["cust_name"].values[0]}',
                    color='crimson')

    ax.set_title(f'Pembelian dan Retur Terbesar - {month.strftime("%B %Y")}')
    ax.set_xlabel('Jumlah Transaksi')
    ax.set_yticks([0, 1])
    ax.set_yticklabels(['Pembelian Terbesar', 'Retur Terbesar'])
    ax.legend()

    def add_labels(bars):
        for bar in bars:
            width = bar.get_width()
            ax.text(width, 
                   bar.get_y() + bar.get_height() / 2, 
                   f'{int(width)}', 
                   ha='left', va='center',
                   fontsize=10)

    add_labels(bars1)
    add_labels(bars2)

if n_months < len(axes):
    for idx in range(n_months, len(axes)):
        fig.delaxes(axes[idx])

plt.tight_layout()
plt.show()
```
